# Algerian Weather Stations Fetcher

This script retrieves the complete list of weather stations located in Algeria using the Meteostat API and exports the data to a CSV file for further analysis.

```python
!pip install meteostat==1.7.6

from meteostat import Stations

stations = Stations()

algerian_stations = stations.region(country='DZ')
station_data = algerian_stations.fetch()
print(station_data.head())

station_data.to_csv(r'algerian_stations.csv')
```
